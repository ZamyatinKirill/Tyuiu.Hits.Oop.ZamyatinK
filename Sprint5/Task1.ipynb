<h1 style="color:DodgerBlue">HashSet<T></h1>



`HashSet<T>` в C# — это коллекция, которая хранит уникальные элементы без определенного порядка. Она полезна, когда нужно быстро проверять наличие элемента или добавлять только уникальные элементы. Давайте рассмотрим, как с ней работать на примере классов `Transport`, `Car`, и `Bike`.

### Создание классов Transport, Car, и Bike

Сначала определим базовый класс `Transport` и его наследники `Car` и `Bike`:

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        // Простая проверка равенства базируется на модели и производителе
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        // Создание хеш-кода на основе модели и производителя
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Number of Doors: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Has Carrier: {HasCarrier}");
    }
}
```

### Работа с HashSet<T>

Теперь рассмотрим, как использовать `HashSet<T>` для хранения объектов `Transport`.

```csharp
using System;
using System.Collections.Generic;


// Создаем HashSet для уникальных объектов транспортных средств
HashSet<Transport> transportSet = new HashSet<Transport>();

// Добавляем объекты Car и Bike в HashSet
var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
 var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

// Попробуем добавить повторяющийся объект
var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);

Console.WriteLine($"Was duplicate car added? {isAdded}"); // false

// Перебираем и выводим информацию обо всех уникальных транспортных средствах
foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

```

### Объяснения и комментарии:

- **Уникальность объектов**: `HashSet<T>` не допускает дублирования. Если вы пытаетесь добавить элемент, который уже присутствует (определяется методом `Equals` и `GetHashCode`), он не добавляется.

- **Методы `Equals` и `GetHashCode`**: Эти методы переопределены для корректной работы с `HashSet`. `Equals` определяет, когда два объекта считаются равными, а `GetHashCode` возвращает хеш-код, используемый для быстрого поиска элемента.

- **Быстрая проверка на наличие**: Благодаря хешированию, `HashSet` позволяет быстро проверять, существует ли элемент в коллекции (`O(1)` — амортизированное время).

- **Отсутствие порядка**: Элементы в `HashSet` не имеют определенного порядка, в отличие от `List<T>`.

`HashSet<T>` идеально подходит для сценариев, где требуется исключительно уникальность элементов и быстрое выполнение операций поиска.

````
using System;
using System.Collections.Generic;

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public override bool Equals(object obj)
    {
        if (obj is Transport transport)
        {
            return Model == transport.Model && Manufacturer == transport.Manufacturer;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer);
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет багажник: {HasCarrier}");
    }
}

HashSet<Transport> transportSet = new HashSet<Transport>();

var car1 = new Car("Model S", "Tesla", 4);
var car2 = new Car("Model 3", "Tesla", 4);
var bike1 = new Bike("Mountain Explorer", "Giant", true);

transportSet.Add(car1);
transportSet.Add(car2);
transportSet.Add(bike1);

var duplicateCar = new Car("Model S", "Tesla", 4);
bool isAdded = transportSet.Add(duplicateCar);
Console.WriteLine($"Был ли добавлен дубликат: {isAdded}");
Console.WriteLine($"Элементов в множестве: {transportSet.Count}");
Console.WriteLine($"Содержит Model 3: {transportSet.Contains(new Car("Model 3", "Tesla", 2))}");

foreach (var transport in transportSet)
{
    transport.DisplayInfo();
    Console.WriteLine();
}

transportSet.Remove(car2);
Console.WriteLine($"После удаления осталось: {transportSet.Count}");
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте при помощи  HashSet. 

````
using System;
using System.Collections.Generic;

public abstract class Animal
{
    private string _name;
    private int _age;

    public HashSet<string> Foods { get; } = new HashSet<string>(StringComparer.OrdinalIgnoreCase);

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя животного не может быть пустым!");
            _name = value;
        }
    }

    public int Age
    {
        get => _age;
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    protected Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public abstract void MakeSound();

    public bool AddFood(string food)
    {
        return Foods.Add(food);
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal other)
        {
            return GetType() == other.GetType()
                && string.Equals(Name, other.Name, StringComparison.OrdinalIgnoreCase);
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(GetType(), Name.ToLowerInvariant());
    }

    public override string ToString()
    {
        return $"{GetType().Name} {Name} ({Age})";
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"{GetType().Name}: {Name}, возраст: {Age}");
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }
}

public class Cat : Animal
{
    public string FurColor { get; set; }

    public Cat(string name, int age, string furColor) : base(name, age)
    {
        FurColor = furColor;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет шерсти: {FurColor}");
    }
}

public class Bird : Animal
{
    public double WingSpan { get; set; }

    public Bird(string name, int age, double wingSpan) : base(name, age)
    {
        WingSpan = wingSpan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поет: Чирик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Размах крыльев: {WingSpan} см");
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }

    public Fish(string name, int age, string waterType) : base(name, age)
    {
        WaterType = waterType;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} молчит");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }
}

public class AnimalNameComparer : IEqualityComparer<Animal>
{
    public bool Equals(Animal x, Animal y)
    {
        if (ReferenceEquals(x, y)) return true;
        if (x is null || y is null) return false;
        return string.Equals(x.Name, y.Name, StringComparison.OrdinalIgnoreCase);
    }

    public int GetHashCode(Animal obj)
    {
        return obj.Name.ToLowerInvariant().GetHashCode();
    }
}

public static class SetPrinter
{
    public static void Print(string title, IEnumerable<Animal> items)
    {
        List<string> names = new List<string>();
        foreach (Animal entry in items)
        {
            names.Add(entry.ToString());
        }
        Console.WriteLine($"{title}: {(names.Count == 0 ? "пусто" : string.Join("; ", names))}");
    }
}

Console.WriteLine("=== Добавление и уникальность ===");
HashSet<Animal> pets = new HashSet<Animal>();
Console.WriteLine($"Рекс добавлен: {pets.Add(new Dog("Рекс", 5, "Овчарка"))}");
Console.WriteLine($"Мурка добавлена: {pets.Add(new Cat("Мурка", 3, "Серый"))}");
Console.WriteLine($"Кеша добавлен: {pets.Add(new Bird("Кеша", 2, 30.5))}");
Console.WriteLine($"Дубликат Рекса добавлен: {pets.Add(new Dog("рекс", 7, "Лайка"))}");
Console.WriteLine($"Собака Мурка добавлена: {pets.Add(new Dog("Мурка", 1, "Такса"))}");
Console.WriteLine($"Элементов в множестве: {pets.Count}");
SetPrinter.Print("Домашние животные", pets);

Console.WriteLine("\n=== Contains, TryGetValue, Remove ===");
Console.WriteLine($"Есть кошка Мурка: {pets.Contains(new Cat("мурка", 0, "Любой"))}");
Console.WriteLine($"Есть кошка Рекс: {pets.Contains(new Cat("Рекс", 0, "Любой"))}");
if (pets.TryGetValue(new Dog("РЕКС", 0, "Любая"), out Animal actualDog))
{
    Console.WriteLine("Найден оригинальный объект:");
    actualDog.DisplayInfo();
}
Console.WriteLine($"Кеша удален: {pets.Remove(new Bird("Кеша", 0, 0))}");
Console.WriteLine($"Кеша удален повторно: {pets.Remove(new Bird("Кеша", 0, 0))}");
SetPrinter.Print("После удаления", pets);

Console.WriteLine("\n=== Операции над множествами ===");
HashSet<Animal> farm = new HashSet<Animal>
{
    new Dog("Рекс", 5, "Овчарка"),
    new Dog("Бобик", 8, "Дворняга"),
    new Bird("Гуся", 4, 90.0),
    new Fish("Немо", 1, "Морская")
};

HashSet<Animal> union = new HashSet<Animal>(pets);
union.UnionWith(farm);
SetPrinter.Print("Объединение", union);

HashSet<Animal> intersection = new HashSet<Animal>(pets);
intersection.IntersectWith(farm);
SetPrinter.Print("Пересечение", intersection);

HashSet<Animal> difference = new HashSet<Animal>(farm);
difference.ExceptWith(pets);
SetPrinter.Print("Только на ферме", difference);

HashSet<Animal> symmetric = new HashSet<Animal>(pets);
symmetric.SymmetricExceptWith(farm);
SetPrinter.Print("Симметрическая разность", symmetric);

Console.WriteLine($"Домашние пересекаются с фермой: {pets.Overlaps(farm)}");
Console.WriteLine($"Пересечение - подмножество фермы: {intersection.IsSubsetOf(farm)}");
Console.WriteLine($"Объединение - надмножество фермы: {union.IsSupersetOf(farm)}");
Console.WriteLine($"Ферма равна домашним: {farm.SetEquals(pets)}");

Console.WriteLine("\n=== RemoveWhere ===");
int removedCount = farm.RemoveWhere(candidate => candidate.Age > 6);
Console.WriteLine($"Удалено старых животных: {removedCount}");
SetPrinter.Print("Ферма после очистки", farm);

Console.WriteLine("\n=== Пользовательский компаратор по имени ===");
HashSet<Animal> byName = new HashSet<Animal>(new AnimalNameComparer());
Console.WriteLine($"Собака Шарик: {byName.Add(new Dog("Шарик", 3, "Спаниель"))}");
Console.WriteLine($"Кошка Шарик: {byName.Add(new Cat("Шарик", 2, "Белый"))}");
Console.WriteLine($"Птица Чижик: {byName.Add(new Bird("Чижик", 1, 18.0))}");
SetPrinter.Print("Множество с компаратором", byName);

HashSet<Animal> byTypeAndName = new HashSet<Animal>();
Console.WriteLine($"Собака Шарик (обычное множество): {byTypeAndName.Add(new Dog("Шарик", 3, "Спаниель"))}");
Console.WriteLine($"Кошка Шарик (обычное множество): {byTypeAndName.Add(new Cat("Шарик", 2, "Белый"))}");

Console.WriteLine("\n=== HashSet<string> внутри животных ===");
Dog buddy = new Dog("Бадди", 4, "Лабрадор");
Console.WriteLine($"мясо: {buddy.AddFood("мясо")}");
Console.WriteLine($"Мясо повторно: {buddy.AddFood("Мясо")}");
Console.WriteLine($"сухой корм: {buddy.AddFood("сухой корм")}");
Console.WriteLine($"Любимая еда Бадди: {string.Join(", ", buddy.Foods)}");

HashSet<string> speciesSet = new HashSet<string>();
foreach (Animal member in union)
{
    speciesSet.Add(member.GetType().Name);
}
Console.WriteLine($"Уникальные виды: {string.Join(", ", speciesSet)}");

farm.Clear();
Console.WriteLine($"После Clear на ферме: {farm.Count}");
````
