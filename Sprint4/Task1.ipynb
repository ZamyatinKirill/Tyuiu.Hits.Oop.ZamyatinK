<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

<h4 style="color:DodgerBlue">Для проверки напишите пример кода на основе классов Vehicle, Car и Motorcycle ниже в блоке с применением интерфейса и абстактного класса:</h4>

````
using System;

public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }

    public void Accelerate(int amount)
    {
        currentSpeed += amount;
        Console.WriteLine($"{Make} {Model} разгоняется до {currentSpeed} км/ч.");
    }
}

public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();
    public abstract void Stop();

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

public class Truck : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} грузовик запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} грузовик остановлен.");
    }
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start();
Console.WriteLine($"Скорость: {myCar.GetCurrentSpeed()} км/ч");
myCar.Stop();

Car realCar = new Car { Make = "Honda", Model = "Civic" };
realCar.Accelerate(40);
Console.WriteLine($"Скорость: {realCar.GetCurrentSpeed()} км/ч");
((IVehicle)realCar).Start();
(realCar as IVehicle).Stop();

IVehicle truck = new Truck { Make = "Volvo", Model = "FH" };
truck.Start();
truck.Stop();

IMotorizedVehicle bike = new Motorcycle { Make = "Ducati", Model = "Monster" };
bike.Start();
bike.RevEngine();
bike.Stop();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
using System;
using System.Collections.Generic;

public interface IAnimal
{
    string Name { get; set; }
    int Age { get; set; }
    void MakeSound();
    void Eat(string food);
    int GetEnergy();
}

public interface IRunner
{
    void Move();
}

public interface ISwimmer
{
    void Move();
}

public interface IFlyer
{
    void Fly();
}

public interface IPet : IAnimal
{
    string Owner { get; set; }
    void Play();
}

public abstract class Animal : IAnimal
{
    private string _name;
    private int _age;
    protected int energy = 50;

    public string Name
    {
        get => _name;
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя животного не может быть пустым!");
            _name = value;
        }
    }

    public int Age
    {
        get => _age;
        set
        {
            if (value < 0)
                throw new ArgumentOutOfRangeException("Возраст не может быть отрицательным!");
            _age = value;
        }
    }

    protected Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public abstract void MakeSound();

    public virtual void Eat(string food)
    {
        energy += 10;
        Console.WriteLine($"{Name} ест {food}. Энергия: {energy}");
    }

    public int GetEnergy()
    {
        return energy;
    }
}

public class Dog : Animal, IPet, IRunner, ISwimmer
{
    public string Breed { get; set; }
    public string Owner { get; set; }

    public Dog(string name, int age, string breed, string owner) : base(name, age)
    {
        Breed = breed;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} лает: Гав-гав!");
    }

    void IPet.Play()
    {
        energy -= 15;
        Console.WriteLine($"{Name} играет с хозяином {Owner}. Энергия: {energy}");
    }

    void IRunner.Move()
    {
        energy -= 10;
        Console.WriteLine($"{Name} бежит. Энергия: {energy}");
    }

    void ISwimmer.Move()
    {
        energy -= 12;
        Console.WriteLine($"{Name} плывет. Энергия: {energy}");
    }
}

public class Cat : Animal, IPet, IRunner
{
    public string FurColor { get; set; }
    public string Owner { get; set; }

    public Cat(string name, int age, string furColor, string owner) : base(name, age)
    {
        FurColor = furColor;
        Owner = owner;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу!");
    }

    void IPet.Play()
    {
        energy -= 5;
        Console.WriteLine($"{Name} играет с клубком. Энергия: {energy}");
    }

    void IRunner.Move()
    {
        energy -= 6;
        Console.WriteLine($"{Name} крадется. Энергия: {energy}");
    }
}

public class Bird : Animal, IFlyer
{
    public double WingSpan { get; set; }

    public Bird(string name, int age, double wingSpan) : base(name, age)
    {
        WingSpan = wingSpan;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поет: Чирик-чирик!");
    }

    void IFlyer.Fly()
    {
        energy -= 8;
        Console.WriteLine($"{Name} летит, размах крыльев {WingSpan} см. Энергия: {energy}");
    }
}

public class Duck : Animal, IFlyer, ISwimmer, IRunner
{
    public Duck(string name, int age) : base(name, age)
    {
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} крякает: Кря-кря!");
    }

    void IFlyer.Fly()
    {
        energy -= 10;
        Console.WriteLine($"{Name} летит над водой. Энергия: {energy}");
    }

    void ISwimmer.Move()
    {
        energy -= 5;
        Console.WriteLine($"{Name} плавает. Энергия: {energy}");
    }

    void IRunner.Move()
    {
        energy -= 7;
        Console.WriteLine($"{Name} переваливается по земле. Энергия: {energy}");
    }
}

Dog dog = new Dog("Рекс", 5, "Овчарка", "Иван");
Cat cat = new Cat("Мурка", 3, "Серый", "Мария");
Bird bird = new Bird("Кеша", 2, 30.5);
Duck duck = new Duck("Дональд", 4);

dog.MakeSound();
dog.Eat("мясо");

IPet dogAsPet = dog;
dogAsPet.Play();
((IRunner)dog).Move();
((ISwimmer)dog).Move();

IPet catAsPet = cat;
catAsPet.MakeSound();
catAsPet.Play();
IRunner catRunner = cat;
catRunner.Move();

IFlyer birdFlyer = bird;
birdFlyer.Fly();
bird.MakeSound();

IFlyer duckFlyer = duck;
ISwimmer duckSwimmer = duck;
IRunner duckRunner = duck;
duckFlyer.Fly();
duckSwimmer.Move();
duckRunner.Move();
duck.MakeSound();

List<IAnimal> animals = new List<IAnimal> { dog, cat, bird, duck };
foreach (IAnimal animal in animals)
{
    Console.WriteLine($"{animal.Name}: энергия {animal.GetEnergy()}");
    if (animal is IFlyer flyer)
        flyer.Fly();
    if (animal is IRunner runner)
        runner.Move();
    if (animal is ISwimmer swimmer)
        swimmer.Move();
}
````
