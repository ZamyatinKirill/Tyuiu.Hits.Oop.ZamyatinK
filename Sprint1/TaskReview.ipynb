```
using System;
using System.Collections.Generic;


public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get => name;
        set => name = value;
    }

    public decimal Price
    {
        get => price;
        set => price = value >= 0 ? value : 0;
    }

    public int Quantity
    {
        get => quantity;
        set => quantity = value >= 0 ? value : 0;
    }

    public string Category
    {
        get => category;
        set => category = value;
    }

    public string Description
    {
        get => description;
        set => description = value;
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
        if (quantity < 0)
        {
            quantity = 0;
        }
    }

    public string GetProductInfo()
    {
        return $"Товар: {name} | Категория: {category} | Цена: {price:C} | Кол-во: {quantity} | Описание: {description} | Общая стоимость: {GetTotalPrice():C}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get => name;
        set => name = value;
    }

    public string EmployeeId
    {
        get => employeeId;
        set => employeeId = value;
    }

    public decimal Salary
    {
        get => salary;
        set => salary = value >= 0 ? value : 0;
    }

    public string ContactInfo
    {
        get => contactInfo;
        set => contactInfo = value;
    }

    public List<Product> Products => products;

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (products.Contains(product))
        {
            if (product.Quantity >= quantity)
            {
                product.UpdateQuantity(-quantity);
                Console.WriteLine($"Продавец {name} продал {quantity} шт. товара \"{product.Name}\".");
            }
            else
            {
                Console.WriteLine($"Недостаточно товара \"{product.Name}\" на складе! Запрошено: {quantity}, доступно: {product.Quantity}");
            }
        }
        else
        {
            Console.WriteLine($"Товар \"{product.Name}\" не закреплен за продавцом {name}.");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name} (ID: {employeeId}) | Оклад: {salary:C} | Контакты: {contactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public string StoreName
    {
        get => storeName;
        set => storeName = value;
    }

    public string Location
    {
        get => location;
        set => location = value;
    }

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Список товаров в магазине \"{storeName}\"");
        bool hasProducts = false;

        foreach (var seller in sellers)
        {
            foreach (var product in seller.Products)
            {
                Console.WriteLine(product.GetProductInfo());
                hasProducts = true;
            }
        }

        if (!hasProducts)
        {
            Console.WriteLine("В магазине пока нет доступных товаров.");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: \"{storeName}\" | Адрес: {location} | Режим работы: {storeHours} | Кол-во продавцов: {sellers.Count}";
    }
}


Product product1 = new Product(
    name: "Ноутбук",
    price: 85000.00m,
    quantity: 8,
    category: "Электроника",
    description: "15.6\" Ноутбук 16GB RAM, SSD 512GB"
);

Seller seller1 = new Seller(
    name: "Алексей Смирнов",
    employeeId: "EMP-102",
    salary: 60000.00m,
    contactInfo: "alexey@store.ru, +7 (900) 123-45-67"
);

seller1.AddProduct(product1);

Store store1 = new Store(
    storeName: "ТехноРай",
    location: "г. Москва, ул. Тверская, д. 15",
    storeHours: "09:00 - 21:00"
);


store1.AddSeller(seller1);

Console.WriteLine("Информация об объектах");
Console.WriteLine(store1.GetStoreInfo());
Console.WriteLine(seller1.GetSellerInfo());
Console.WriteLine(product1.GetProductInfo());

Console.WriteLine("\nИнвентаризация магазина");
store1.ListProducts();

Console.WriteLine("\nПроцесс продажи");
seller1.SellProduct(product1, 3);

Console.WriteLine("\nСостояние товара после продажи");
Console.WriteLine(product1.GetProductInfo());

Console.WriteLine("\nПополнение склада (+5 шт.):");
product1.UpdateQuantity(5);
Console.WriteLine($"Новое количество: {product1.Quantity} шт.");
Console.WriteLine($"Новая общая стоимость товара: {product1.GetTotalPrice():C}");

Console.WriteLine("\nПопытка установить отрицательную цену и количество:");
product1.Price = -500;
product1.Quantity = -10;
Console.WriteLine($"Цена после попытки задать -500: {product1.Price:C}");
Console.WriteLine($"Количество после попытки задать -10: {product1.Quantity}");

product1.Price = 85000.00m;
product1.Quantity = 10;

Console.WriteLine("\nДобавление второго продавца и товара:");
Product product2 = new Product("Мышь беспроводная", 2500.00m, 20, "Аксессуары", "Оптическая 1600 DPI");
Seller seller2 = new Seller("Елена Петрова", "EMP-105", 55000.00m, "elena@store.ru");

seller2.AddProduct(product2);
store1.AddSeller(seller2);

Console.WriteLine(seller2.GetSellerInfo());
Console.WriteLine(store1.GetStoreInfo());

Console.WriteLine("\nПопытка продать 50 шт. мышей (на складе всего 20):");
seller2.SellProduct(product2, 50);

Console.WriteLine("\nУспешная продажа 5 шт. мышей:");
seller2.SellProduct(product2, 5);
Console.WriteLine($"Остаток мышей на складе: {product2.Quantity} шт.");

Console.WriteLine("\nИтоговая инвентаризация магазина:");
store1.ListProducts();
```
